Memeriksa nilai ekstrem

```python
# Menentukan kolom yang akan diperiksa
kolom_outlier = [
    'exposure_score',
    'tasks_total',
    'tasks_ai_alone',
    'tasks_ai_with_software',
    'tasks_not_exposed',
    'tasks_seen_in_ai_use',
    'us_employment',
    'median_annual_pay_usd'
]

# Menghitung outlier menggunakan metode IQR
for kolom in kolom_outlier:
    Q1 = df[kolom].quantile(0.25)
    Q3 = df[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    outlier = df[
        (df[kolom] < batas_bawah) |
        (df[kolom] > batas_atas)
    ]

    print(f"\nKolom: {kolom}")
    print("Jumlah outlier:", len(outlier))
    print("Batas bawah:", batas_bawah)
    print("Batas atas:", batas_atas)
```

Data Transformation (Transformasi Data)

*Transformasi Jumlah Tugas menjadi Persentase*

```python
# Mengubah jumlah tugas menjadi persentase
df_preprocessing['persentase_ai_alone'] = (
    df_preprocessing['tasks_ai_alone'] /
    df_preprocessing['tasks_total'] * 100
)

df_preprocessing['persentase_ai_with_software'] = (
    df_preprocessing['tasks_ai_with_software'] /
    df_preprocessing['tasks_total'] * 100
)

df_preprocessing['persentase_not_exposed'] = (
    df_preprocessing['tasks_not_exposed'] /
    df_preprocessing['tasks_total'] * 100
)
```

```python
df_preprocessing[
    [
        'occupation',
        'tasks_total',
        'tasks_ai_alone',
        'tasks_ai_with_software',
        'tasks_not_exposed',
        'persentase_ai_alone',
        'persentase_ai_with_software',
        'persentase_not_exposed'
    ]
].head(10)
```

*Validasi Hasil Transformasi Persentase*

```python
# Menjumlahkan seluruh persentase pembagian tugas
df_preprocessing['total_persentase'] = (
    df_preprocessing['persentase_ai_alone'] +
    df_preprocessing['persentase_ai_with_software'] +
    df_preprocessing['persentase_not_exposed']
)

# Memeriksa apakah total persentase mendekati 100%
jumlah_tidak_100 = (
    ~df_preprocessing['total_persentase'].round(6).eq(100)
).sum()

print("Jumlah data dengan total persentase bukan 100%:", jumlah_tidak_100)
```

*Normalisasi Data Numerik (Min-Max Normalization)*

```python
# Normalisasi jumlah tenaga kerja
df_preprocessing['employment_normalized'] = (
    (df_preprocessing['us_employment'] - df_preprocessing['us_employment'].min()) /
    (df_preprocessing['us_employment'].max() - df_preprocessing['us_employment'].min())
)

# Normalisasi gaji tahunan
df_preprocessing['salary_normalized'] = (
    (df_preprocessing['median_annual_pay_usd'] - df_preprocessing['median_annual_pay_usd'].min()) /
    (df_preprocessing['median_annual_pay_usd'].max() - df_preprocessing['median_annual_pay_usd'].min())
)
```

```python
df_preprocessing[
    [
        'occupation',
        'us_employment',
        'employment_normalized',
        'median_annual_pay_usd',
        'salary_normalized'
    ]
].head(10)
```

*Validasi Hasil Normalisasi*

```python
# Memeriksa nilai minimum dan maksimum hasil normalisasi
df_preprocessing[
    ['employment_normalized', 'salary_normalized']
].agg(['min', 'max'])
```

*Encoding Kategori*

```python
# Mapping tingkat paparan AI
mapping_exposure = {
    'Low': 1,
    'Moderate': 2,
    'Elevated': 3,
    'High': 4,
    'Very high': 5
}

# Melakukan encoding kategori
df_preprocessing['exposure_level_encoded'] = (
    df_preprocessing['exposure_level'].map(mapping_exposure)
)
```

*Encoding Kategori field (One-Hot Encoding)*

```python
# Melakukan One-Hot Encoding pada kolom field
df_preprocessing = pd.get_dummies(
    df_preprocessing,
    columns=['field'],
    prefix='field',
    dtype=int
)
```

*Validasi One-Hot Encoding*

```python
kolom_field_encoded = [
    kolom for kolom in df_preprocessing.columns
    if kolom.startswith('field_')
]

jumlah_field_per_baris = df_preprocessing[kolom_field_encoded].sum(axis=1)

print("Jumlah baris dengan lebih dari 1 kategori:",
      (jumlah_field_per_baris > 1).sum())

print("Jumlah baris dengan 0 kategori:",
      (jumlah_field_per_baris == 0).sum())
```

```python
df_preprocessing[kolom_field_encoded].sum().sort_values(ascending=False)
```

*Validasi One-Hot Encoding (Pemeriksaan Akhir)*

```python
print("Jumlah baris dengan lebih dari 1 kategori:",
      (jumlah_field_per_baris > 1).sum())

print("Jumlah baris dengan 0 kategori:",
      (jumlah_field_per_baris == 0).sum())
```
